General Questions

```python
import numpy as np

# Columns: [Age, Math Marks, Science Marks]
data = np.array([
    [18, 85, 78],   # Student 1
    [19, 92, 88],   # Student 2
    [17, 76, 95],   # Student 3
    [18, 65, 70],   # Student 4
    [20, 90, 85]    # Student 5
])
data
```

#Get the shape of the matrix.

```python
data.shape
```


#Find the average age of students.


```python
avg_age = data[:, 0].mean()
print("Average Age:", avg_age)
```

#Extract Math marks of all students.

```python
data[:,1]
```

#Find the highest Science mark.

```python
data[:,2].max()
```

#Get details of the student who scored more than 90 in Math.

```python
a = data[data[:,1] > 90]
a
```

#Increase Math marks of all students by 5.

```python
data[:,1] = data[:,1] + 5
print(data[:,1])
```

#Find how many students are younger than 19.

```python
a = data[data[:,0] < 19 , 0]
print(a,"there are :",a.size,"students younger then 19 ")
```

#Calculate the average marks in each subject (column-wise mean).


```python
arr1 = data[:,1].mean()
print("math = ",arr1)
arr2 = data[:,2].mean()
print("science = ",arr2)
```

#Get data of students who scored at least 80 in both subjects.

```python
score = data[(data[:,1] >= 80) & (data[:,2] >= 80)]
print("student scored at least 80 = ",score)
```

#Replace all Science marks < 75 with 0.

```python
data[data[:,2] < 75,2] = 0
print(data)
```
